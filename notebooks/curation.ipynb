# JEDI3sub Event Curation

Set the data path once, then use the manual, fast-event, and slow-event sections independently. For a shared `Data/` path, choose one animal and experiment before choosing a scan/domain; only that experiment's metadata is read. Each section writes its own JSON file inside the selected experiment's `PF/.curation/` folder.

```python
from pathlib import Path

from vnoiser.curation import EventCurationDashboard

# Point this at the shared Data folder, one experiment/PF folder, or a raw .mat file.
DATA_PATH = Path(
    "../data"
)
```

## 1. Manual Curation

Thresholds the denoised trace to find candidates and starts with no template. Panel A1 uses a data-calibrated range, saves its value per recording, and changes the candidate threshold. Events marked **Yes** become the evolving template; **No** events are excluded from the second-pass preview. Manual decisions remain available across threshold changes until explicitly cleared.

```python
manual_dashboard = EventCurationDashboard(
    data_path=DATA_PATH,
    mode="manual",
    duration_s=None,
    auto_load=False,
)
manual_dashboard.display()
```

## 2. Fast-Event Seeded Curation

Thresholds the denoised trace shown in black, then seeds the template from the top 25% highest-amplitude candidates. Panel A1 uses a data-calibrated range, saves its value per recording, and recalculates candidates, the seed template, cosine scores, and downstream panels. Automatic pass/reject colors remain provisional until a manual **Yes** or **No** decision is saved; manual decisions persist across threshold changes until explicitly cleared.

```python
fast_dashboard = EventCurationDashboard(
    data_path=DATA_PATH,
    mode="fast",
    duration_s=None,
    auto_load=False,
)
fast_dashboard.display()
```

## 3. Slow-Event Seeded Curation (<40 Hz)

Applies a zero-phase `<40 Hz` low-pass filter directly to the denoised trace, thresholds the blue filtered curve, and seeds its template from the top 25% highest-amplitude slow candidates. Panel A shows the original denoised trace in black and the filtered analysis trace in blue; Panel A1 uses a data-calibrated range and saves its threshold per recording. Its labels and template never overwrite the fast-event state, and manual decisions persist across threshold changes until explicitly cleared.

```python
slow_dashboard = EventCurationDashboard(
    data_path=DATA_PATH,
    mode="slow",
    duration_s=None,
    slow_cutoff_hz=40.0,
    auto_load=False,
)
slow_dashboard.display()
```

The three saved states are `manual_template_curation.json`, `fast_template_curation.json`, and `slow_template_curation.json`. Reopening this notebook and loading the same scan/domain restores each section from its own file.
